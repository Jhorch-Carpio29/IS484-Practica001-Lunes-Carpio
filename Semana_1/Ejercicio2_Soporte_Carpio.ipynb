## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Componente | Descripción |
| :--- | :--- |
| **Percepción (S)** | Variables del ticket: `tiempo_espera_minutos` (entero), `nivel_urgencia` (`"baja"`, `"media"`, `"alta"`) y `cliente_premium` (booleano). |
| **Acciones (A)** | Niveles de escalamiento: `"resolver en Nivel 1"`, `"escalar a Nivel 2"` o `"escalar a Nivel 3"`. |
| **Entorno (E)** | Sistema digital de mesa de ayuda (Help Desk) y gestión de tickets de soporte técnico. |
| **Objetivo** | Asignar cada ticket al nivel de soporte adecuado para cumplir con los acuerdos de nivel de servicio (SLA) y optimizar la carga del equipo. |
| **Medida de desempeño (P)** | Minimizar el tiempo medio de resolución (MTTR), reducir el reescalamiento innecesario a especialistas y maximizar el índice de satisfacción del cliente (CSAT). |

### Justificación de las reglas

Priorizar tickets con urgencia alta o de clientes premium garantiza el cumplimiento de los contratos SLA críticos para el negocio. Acumular tiempos de espera prolongados incrementa la insatisfacción del usuario, justificando el escalamiento automático a niveles superiores incluso ante urgencias inicialmente bajas. Resolver solicitudes de baja urgencia y corto tiempo en Nivel 1 preserva la disponibilidad del personal altamente especializado de Niveles 2 y 3 para incidencias complejas.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    """
    Evalúa el destino de escalamiento de un ticket según tiempo de espera, urgencia y categoría del cliente.
    Retorna una tupla: (accion, motivo)
    """
    urgencia = nivel_urgencia.lower()
    
    # Regla 1: Escalamiento crítico a Nivel 3
    if urgencia == "alta" and (cliente_premium or tiempo_espera_minutos >= 30):
        return "escalar a Nivel 3", f"Urgencia alta con prioridad crítica (Premium: {cliente_premium}, Espera: {tiempo_espera_minutos} min)."
    
    # Regla 2: Escalamiento intermedio a Nivel 2
    if urgencia == "alta":
        return "escalar a Nivel 2", f"Urgencia alta requiere atención especializada (Espera: {tiempo_espera_minutos} min)."
    if urgencia == "media" and (cliente_premium or tiempo_espera_minutos >= 20):
        return "escalar a Nivel 2", f"Urgencia media con priorización (Premium: {cliente_premium}, Espera: {tiempo_espera_minutos} min)."
    if urgencia == "baja" and cliente_premium and tiempo_espera_minutos >= 40:
        return "escalar a Nivel 2", f"Cliente premium con espera prolongada ({tiempo_espera_minutos} min) en urgencia baja."
    if tiempo_espera_minutos >= 60:
        return "escalar a Nivel 2", f"Tiempo de espera acumulado ({tiempo_espera_minutos} min) excede el SLA de Nivel 1."
    
    # Regla 3: Atención básica en Nivel 1
    return "resolver en Nivel 1", f"Caso estándar asignado a Nivel 1 (Urgencia: {urgencia}, Espera: {tiempo_espera_minutos} min)."
```

### Simulación y pruebas

```python
# 6 combinaciones con condiciones cruzadas
casos_prueba_soporte = [
    # (tiempo_espera_minutos, nivel_urgencia, cliente_premium)
    (5, "alta", True),     # Urgencia alta + Premium -> Escalar a Nivel 3
    (5, "alta", False),    # Urgencia alta + No Premium -> Escalar a Nivel 2
    (10, "media", False),  # Urgencia media + Tiempo corto -> Resolver en Nivel 1
    (25, "media", False),  # Urgencia media + Tiempo moderado -> Escalar a Nivel 2
    (50, "baja", True),    # Urgencia baja + Premium + Tiempo alto -> Escalar a Nivel 2
    (10, "baja", False),   # Urgencia baja + No Premium + Tiempo corto -> Resolver en Nivel 1
]

print("=== EVALUACIÓN DE TICKETS DE SOPORTE ===")
for i, (espera, urgencia, premium) in enumerate(casos_prueba_soporte, 1):
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(f"Ticket {i} [Espera: {espera}m, Urgencia: '{urgencia}', Premium: {premium}]")
    print(f"  └─ Acción: [{accion.upper()}] -> Motivo: {motivo}\n")
```
