## 5.2. Parte A: Implementación de DFS

A diferencia de BFS (que explora nivel por nivel usando una Cola FIFO con `pop(0)`), el algoritmo **DFS (Depth-First Search)** utiliza una **Pila LIFO** (*Last In, First Out*) mediante `.pop()`, lo que causa que explore hasta el fondo de cada rama antes de retroceder.

```python
import numpy as np

# Definición del mapa de la ciudad (Grafo)
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS (Pila LIFO).
    """
    pila = [[inicio]]  # Inicializamos la pila con el camino inicial
    visitados = set()  # Conjunto para evitar ciclos
    
    while pila:
        camino = pila.pop()  # Extrae el ÚLTIMO camino agregado (LIFO)
        nodo_actual = camino[-1]
        
        if nodo_actual == objetivo:
            return camino
        
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            
            for vecino in grafo[nodo_actual]:
                # COMPLETADO: Construimos el nuevo camino extendiéndolo con el vecino
                nuevo_camino = camino + [vecino]
                # COMPLETADO: Lo agregamos a la pila
                pila.append(nuevo_camino)
                
    return None

# Ejecución del algoritmo DFS
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")

print("=== RESULTADO DE BÚSQUEDA DFS ===")
print("Camino encontrado (DFS):", resultado_dfs)
```

### Comparación de Resultados: BFS vs. DFS

* **Camino BFS (Sección 4.3):** `['Almacen', 'Centro', 'ZonaNorte']` (Longitud: 2 tramos).
* **Camino DFS (Sección 5.2):** `['Almacen', 'Terminal', 'Mercado', 'ZonaNorte']` (Longitud: 3 tramos).

**Respuestas:**
1. **¿Son el mismo camino?:** **No**, son caminos totalmente distintos.
2. **¿Cuál tiene menos pasos?:** **BFS tiene menos pasos** (2 tramos frente a los 3 tramos de DFS).
3. **Explicación:** BFS explora en anchura (nivel por nivel de cercanía), garantizando siempre encontrar el camino con la menor cantidad de pasos. En cambio, DFS procesa el último camino ingresado a la pila (LIFO), adentrándose por la rama de `Terminal` $\rightarrow$ `Mercado` $\rightarrow$ `ZonaNorte` y retornando la primera solución encontrada, aunque requiera más tramos.

## 5.3. Parte B: Encontrar la observación más cercana

Dado el dataset de cuatro observaciones de temperatura y humedad, calculamos la distancia euclidiana entre una **nueva observación** (`[20.5°C, 48% humedad]`) y cada una de las existentes para identificar cuál es la más parecida.

```python
# Recreamos el dataset histórico
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])

# Nueva observación a clasificar
nueva_observacion = np.array([20.5, 48])
distancias = []

print("=== DISTANCIAS EUCLIDIANAS A CADA OBSERVACIÓN ===")

for i, obs in enumerate(dataset):
    # COMPLETADO: Cálculo de distancia euclidiana usando la norma L2
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observacion {i} {obs}: {d:.4f}")

# np.argmin nos da el índice del valor con la menor distancia
indice_mas_cercano = np.argmin(distancias)

print("\n=== RESULTADO DE SIMILITUD ===")
print("La observacion mas parecida es la numero:", indice_mas_cercano)
print(f"Valores del registro más cercano: {dataset[indice_mas_cercano]}")
```

## 5.4. Parte C: Reflexión

### 1. ¿Por qué la distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones?
Porque calcula la separación en línea recta dentro de un espacio vectorial geométrico N-dimensional:

$$d(\mathbf{a}, \mathbf{b}) = \sqrt{\sum_{i=1}^{n} (a_i - b_i)^2}$$

Si dos observaciones poseen características numéricas muy similares (temperatura y humedad cercanas), la diferencia entre sus componentes $(a_i - b_i)$ será pequeña, produciendo una distancia reducida. Por lo tanto, **a menor distancia euclidiana, mayor es el grado de similitud**.

---

### 2. ¿En qué casos podría no ser la mejor opción?
Falla cuando las características del dataset se miden en **escalas o magnitudes numéricas muy dispares**.

* **Ejemplo práctico (Soles vs. Años):**
  Si comparamos dos registros con las variables *Sueldo mensual en Soles* y *Edad en Años*:
  * Registro 1: `[S/. 3000, 25 años]`
  * Registro 2: `[S/. 1500, 30 años]`
  
  $$\text{Diferencia de Sueldo}^2 = (3000 - 1500)^2 = 2,250,000$$
  $$\text{Diferencia de Edad}^2 = (25 - 30)^2 = 25$$

* **Consecuencia:** La variable del sueldo domina completamente el resultado de la distancia, anulando el efecto de la edad.
* **Solución:** En estos escenarios es obligatorio aplicar **normalización o estandarización** de datos (como *MinMaxScaler* o *StandardScaler*) antes de calcular la distancia, o utilizar otras métricas como la **distancia de Manhattan** o la **similitud de coseno**.
